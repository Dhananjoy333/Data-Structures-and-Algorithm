Q. Detect if there is a cycle in given graph

Adjacency list :  
0 --> []  
1 --> [2, 4]  
2 --> [1, 3, 6]  
3 --> [2]  
4 --> [1, 5]  
5 --> [4, 7]  
6 --> [2, 7]  
7 --> [5, 6]  

### Using BFS

1. First we create the visited array and a queue to track the nodes.
2. Inside the queue instead of simply storing the node we also store parent node as (node, parent). Idea : If a cycle exist as we traverse we will find a node which is not parent but is visited by some other node. Hence the visited node connect the current node to some other node creating a closed cycle. 
3. For starting node since it won't have parent we simply put -1.

```python
from collections import deque
class CheckCycle_BFS :
    def isCycle(self, v, edges):
        adj_list = [[] for _ in range(v)]
        for u,w in edges :
            adj_list[u].append(w)
            adj_list[w].append(u)
            
        visited = [0] * v
        for i in range(0,v):
            if visited[i] == 1 :
                continue
            queue = deque()
            queue.append((i,-1))
            visited[0] = 1
            while len(queue) != 0:
                node, parent = queue.popleft()
                for adjNode in adj_list[node]:
                    if visited[adjNode] == 0:
                        visited[adjNode] = 1
                        queue.append((adjNode,node))
                    else:
                        if adjNode != parent:
                            return True
        return False

test = CheckCycle_BFS()
print(test.isCycle(4,[[0,1],[0,2],[1,2],[2,3]]))

test2 = CheckCycle_BFS()
print(test2.isCycle(7,[[2,1],[1,0],[0,3],[3,4],[4,6],[6,5],[5,1]]))

# T.C : O(N + 2E)
# S.C : O(N) + O(N)
    
```
