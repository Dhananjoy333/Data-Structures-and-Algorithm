# Dynamic Programming
The four approaches are usually:

1. Recursion (Brute Force) (Top-Down Approach): TC --> O( $2^n$ ) , SC --> O(N)[stack space].
2. Memoization (Top-Down DP) : TC --> O(N) , SC --> O(N)[stack space] + O(N)[DP Array]
3. Tabulation (Bottom-Up DP) : TC --> O(N) , SC --> O(N)[DP Array]
4. Space-Optimized Tabulation : TC --> O(N) , SC --> O(1)

They are not four different DP concepts. They are mostly different implementations of the same DP recurrence.

## Recursion
It is top-down approach which means we start at some value in recursive tree and go down towards base condition.

```python
# Q. Find nth fibonacci number
def fibonacci(n):
    if n <= 1:
        return n
    return(fibonacci(n-1)+fibonacci(n-2))
print(fibonacci(9))

# Time Complexity : O(2^n)
# Space Complexity : O(n)[stack space]
```

## Memoization
To store answers or overlapping subproblems. Whenever in recursive we see sub problem we can use memoization.

<img src="../assets/overlapping_Subproblem.png" width="400px">

With normal recursion we are finding the overlapping subproblem again and again instead we can store the values of already calculated subproblem in an array/list called dp.

```python
def memoization(num):
    dp = [-1]*(num+1)
    def fibonacci(num,dp):
        if num <= 1:
            return num
        if dp[num] != -1:
            return dp[num]
        dp[num] = fibonacci(num-1,dp)+fibonacci(num-2,dp)
        return dp[num]
    return fibonacci(num,dp)
print(memoization(9))
```

## Tabulation [Bottom-Up Approach]
Here we go from base case towards the solution. If a problem can be solved by memoization it can be solved by Tabulation.

```python
def Tabulation(num):
    dp = [-1]*(num+1)
    dp[0] = 0
    dp[1] = 1
    for i in range(2,num+1):
        dp[i] = dp[i-1]+dp[i-2]
    return dp[num]
print(Tabulation(9))
```

## Space-Optimized Tabulation [Bottom-Up Approach]
If in problem to find the solution we don't need all values instead few values in this case to solve for n we simply need n-1 and n-2, hence in such cases space optimization is possible.

```python
def OptiTabulation(num):
    prev2 = 0
    prev = 1
    current = 0
    for _ in range(2,num+1):
        current = prev + prev2
        prev2 = prev
        prev = current
    return prev
print(OptiTabulation(9))
```
